#ñursi

```python
vocal="aeiou"
tildes="áéíóú"
consonante="bcdfghjklmñpqrstvwxyz"
```

```python
#vocales uno a laderecha
#tildes 2 hacia derecha
#cononante 1 hacia la izquierda
#hola-->guke
```

```python
mensaje="ñursi"
```

```python
# @title
codificado=""
for variable in mensaje:
  if variable in vocal:
    match variable:
      case "a": codificado=codificado+"e"
      case "e": codificado=codificado+"i"
      case "i": codificado=codificado+"o"
      case "o": codificado=codificado+"u"
      case "u": codificado=codificado+"a"
  elif variable in tildes:pass
  elif variable in consonante:
    codificado=codificado+consonante[consonante.index(variable)-1]
  else:
    codificado=codificado+variable
print(codificado)
```
